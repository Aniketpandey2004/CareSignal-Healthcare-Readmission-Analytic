# Final Model Evaluation

This notebook reads the outputs from `train_model.py`. It does not create a new random split or retrain a second model, so the evaluation design stays consistent with the patient-grouped training pipeline.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'reports').exists() and (ROOT.parent / 'reports').exists():
    ROOT = ROOT.parent
REPORTS = ROOT / 'reports'

metrics_path = REPORTS / 'patient_level_final_metrics.csv'
comparison_path = REPORTS / 'patient_level_model_comparison.csv'
metadata_path = REPORTS / 'training_metadata.json'

if not metrics_path.exists():
    raise FileNotFoundError('Run notebooks/04_modeling.ipynb first to generate final reports.')

metrics = pd.read_csv(metrics_path)
comparison = pd.read_csv(comparison_path) if comparison_path.exists() else pd.DataFrame()
metadata = json.loads(metadata_path.read_text()) if metadata_path.exists() else {}

print('FINAL PATIENT-LEVEL HOLDOUT METRICS')
display(metrics.T)
print('VALIDATION MODEL COMPARISON')
display(comparison)
print('SPLIT AUDIT')
display(pd.DataFrame([metadata]).T)
```

## Interpretation checklist

- Precision: among encounters flagged by the model, how many were positive in the dataset?
- Recall: among positive encounters in the holdout, how many were flagged?
- PR-AUC: useful for comparing ranking performance when the positive class is uncommon.
- Confusion matrix: inspect false positives and false negatives, not accuracy alone.
- The selected threshold is an exploratory operating point chosen on validation data, not a clinically validated cutoff.
- This retrospective result does not establish performance in a different hospital or current population.
